# A/B Test: Premieres Week
Analysis of block position impact on engagement.

```
import pandas as pd
import scipy.stats as stats
df = pd.read_csv('../data/ab_data.csv')
df.head()
```

```
control = df[df.group=='A']
test = df[df.group=='B']
ctr_control = control.clicked.mean()
ctr_test = test.clicked.mean()
ctr_control, ctr_test
```

```
z_stat, p_value = stats.proportions_ztest(
    [control.clicked.sum(), test.clicked.sum()],
    [len(control), len(test)]
)
p_value
```
